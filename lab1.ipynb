```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

modelo = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

oraciones = [
    "Cien años de soledad narra la historia de la familia Buendía en el pueblo de Macondo.",
    "El principito es un cuento filosófico sobre un niño que viaja por el universo.",
    "El manual de Python avanzado es indispensable para aprender estructuras de datos.",
    "Don Quijote de la Mancha es considerada la obra cumbre de la literatura española.",
    "Los libros de ciencia ficción exploran futuros distópicos y el impacto de la tecnología.",
    "Orgullo y prejuicio relata las complicaciones del amor y la clase social en Inglaterra.",
    "El señor de los anillos es una obra épica de fantasía sobre la destrucción de un anillo mágico.",
    "Nuestra biblioteca universitaria ofrece acceso a miles de revistas científicas y ensayos.",
    "Las novelas de misterio y suspenso siempre logran mantener al lector en tensión.",
    "Fahrenheit 451 presenta un mundo donde los bomberos tienen la misión de quemar libros.",
    "La sombra del viento sigue a un joven en la Barcelona de la posguerra que descubre un libro maldito.",
    "El género de la novela negra se caracteriza por ambientes oscuros e investigaciones criminales.",
    "Un buen libro de algoritmos debe incluir problemas prácticos explicados paso a paso.",
    "Harry Potter es una saga que cuenta las aventuras de un joven mago y sus amigos.",
    "La poesía clásica ayuda a comprender la métrica, el ritmo y la belleza del lenguaje.",
    "Hábitos atómicos enseña métodos prácticos para mejorar la productividad diaria.",
    "El retrato de Dorian Gray trata sobre un hombre cuyo retrato envejece mientras él se mantiene joven.",
    "Las historias de terror psicológico juegan con la mente y la percepción de la realidad del lector.",
    "La metamorfosis comienza cuando el protagonista despierta transformado en un insecto gigante.",
    "Ensayo sobre la ceguera plantea un escenario donde la sociedad pierde la vista repentinamente."
]
consultas_prueba = [
    "Busco recomendaciones de novelas de magia, espadas y aventuras épicas.",
    "Quiero leer historias de detectives, crímenes o suspenso policial.",
    "Textos técnicos para mejorar en programación y resolver problemas matemáticos.",
    "Libros sobre futuros opresivos, censura y tecnología peligrosa.",
    "Obras clásicas de la literatura que traten sobre el romance y la sociedad."
]

embeddings_oraciones = modelo.encode(oraciones)

for consulta in consultas_prueba:
    print(f"\nConsulta: '{consulta}'")
    
    embedding_consulta = modelo.encode([consulta])
    
    similitudes = cosine_similarity(embedding_consulta, embeddings_oraciones)[0]
    
    indice_mejor_coincidencia = np.argmax(similitudes)
    mejor_oracion = oraciones[indice_mejor_coincidencia]
    puntaje = similitudes[indice_mejor_coincidencia]
    
    print(f"Mejor coincidencia: '{mejor_oracion}'")
    print(f"Puntaje de similitud: {puntaje:.4f}")
```
